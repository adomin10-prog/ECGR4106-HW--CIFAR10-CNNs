## Setup and verification run

This cell:

1. clones the GitHub repository,
2. enters the homework project folder,
3. installs `requirements.txt`,
4. runs `run_verification.py`,
5. displays the verification results.

````python
from pathlib import Path
import os
import subprocess
import sys

REPO_URL = "https://github.com/adomin10-prog/ECGR4106-HW--CIFAR10-CNNs.git"
REPO_DIR = Path("/content/ECGR4106-HW--CIFAR10-CNNs")
PROJECT_DIR = REPO_DIR / "ECGR4106-HW1-CIFAR10-CNNs"

if not PROJECT_DIR.exists():
    print("Cloning repository...")
    subprocess.run(["git", "clone", REPO_URL, str(REPO_DIR)], check=True)
else:
    print("Repository already exists.")

os.chdir(PROJECT_DIR)
print("Current project folder:")
print(Path.cwd())

print("\nInstalling requirements...")
subprocess.run([sys.executable, "-m", "pip", "install", "-r", "requirements.txt"], check=True)

print("\nRunning verification...")
subprocess.run([sys.executable, "run_verification.py", "--mode", "verification"], check=True)

print("\nDisplaying verification results...")
import pandas as pd
from IPython.display import display, Image

summary_path = Path("verification_results/verification_summary.csv")
chart_path = Path("verification_results/verification_accuracy_bar_chart.png")

display(pd.read_csv(summary_path))
display(Image(filename=str(chart_path)))
````
